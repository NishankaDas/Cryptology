```python
import math
import numpy as np
```

```python
k = 3
text = "hatti ma tim tim tara mathe pare dim"
```

```python
def encryp(k, text):
    x = []
    for i in range(0,len(text)):
        if i%k == 0:
            u = []
        u.append(text[i])  
        if i%k == 0:
            x.append(u)
    new = np.array(x).T.tolist()   
    h = ''
    for i in new:
        for j in i:
            h = h + j
    return h
```

```python
encryp(k,text)
```

```python
cipher = encryp(k,text)
```

```python
def decryp(k,cipher):
    p = math.ceil(len(cipher)/k)
    x = []
    for i in range(0,len(cipher)):
        if i%p == 0:
            u = []
        u.append(cipher[i])  
        if i%p == 0:
            x.append(u)
    hm =  np.array(x).T.tolist() 
    z = ''
    for i in hm:
        for j in i:
            z = z + j
    return z
```

```python
decryp(k, cipher)
```
