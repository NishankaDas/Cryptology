```python
k = [2,7,9,4,0]
text = "Ami Hati Tumi Jamini"
```

```python
def encryp(key, text):
    cry = ""
    text = text.upper()
    
    for i in range(len(text)):
        if text[i] == " ":
            cry += " " 
        else:
            # Convert character 'A'-'Z' to numerical index 0-25
            p = ord(text[i]) - ord('A')
            
            # Access the integer shift directly from the key array
            k = key[i % len(key)]
            
            # Apply the shift and wrap modulo 26
            c = (p + k) % 26
            
            # Convert back to character and append
            cry += chr(c + ord('A'))
            
    return cry
```

```python
Faaa = encryp(k, text)
```

```python
def decryp(key, text):
    smile = ""
    text = text.upper()
    
    for i in range(len(text)):
        if text[i] == " ":
            smile += " " 
        else:
            # Convert character 'A'-'Z' to numerical index 0-25
            p = ord(text[i]) - ord('A')
            
            # Access the integer shift directly from the key array
            k = key[i % len(key)]
            
            # Apply the shift and wrap modulo 26
            c = (p - k) % 26
            
            # Convert back to character and append
            smile += chr(c + ord('A'))
            
    return smile
```
